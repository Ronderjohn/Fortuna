# Fortuna Backtest Example

Run an EMA crossover backtest on AAPL daily data.

```python
from pathlib import Path

from fortuna.backtesting.engine import BacktestEngine
from fortuna.config.settings import load_settings
from fortuna.data.manager import MarketDataManager
from fortuna.evaluation.ranker import StrategyRanker
from fortuna.strategy.loader import load_strategy
from fortuna.utils.logging import setup_logging

settings = load_settings()
setup_logging(settings.log_level)
root = Path("..").resolve()
```

```python
strategy_path = root / "strategies" / "generated" / "ema_crossover.json"
strategy = load_strategy(strategy_path)
symbol = strategy.symbol
timeframe = strategy.timeframe

mdm = MarketDataManager(settings)
df = mdm.get_ohlcv(symbol, timeframe)
df.tail()
```

```python
engine = BacktestEngine(settings)
result = engine.run(strategy, df, symbol=symbol)
result.metrics
```

```python
ranker = StrategyRanker(settings)
ranked = ranker.rank_one(strategy_path, result)
print(f"Score: {ranked.score.composite:.1f} | Passed: {ranked.score.passed}")
dest = ranker.persist(ranked)
dest
```
